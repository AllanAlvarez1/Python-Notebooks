#File Handling
Python has multiple functions for creating, reading, updating, and deleting files.

The main function for working with files is the 'open()' function. The parameters for it are filename and mode.

There are 4 different modes for opening a file

* 'r' - Read- Default value which opens a file for reading and gives an error if it does not exist
* 'a' - Append- Opens a file for appending and creates the file if it does not exist
* 'w'- Write- Opens a file for writing and creates the file if it does not exist
* 'x' - Create- Creates the specified file and returns an error if the file exists

You can even specify if the file should be handles as binary or text mode using:

* 't' - Text- Default value which is Text mode
* 'b' - Binary - Binary mode which uses images and such

#Syntax
To open a file for reading you must specify the name of the file

```python
myFile = open("Demo.txt")
```

#Read
Opening a file merely makes it ready to use. In order to examine the contents of the file you must use the 'read()' function.

```python
myFile = open("Demo.txt")
print(myFile.read())
```

You can use a file path as well if you need to find it in a specific location on your device.

#Reading Parts of the File
The 'read()' function returns the whole text but you can specify how many characters you want to return

```python
myFile = open("Demo.txt")
print(myFile.read(10))
```

You can return one line by using the readline() method

```python
myFile = open("Demo.txt")
print(myFile.readline())
```

Call it again and you can read the first two lines

```python
myFile = open("Demo.txt")
print(myFile.readline())
print(myFile.readline())
print(myFile.readline())
```

#Loop Reading
Looping through the lines of the file allows you to read the whole file line by line

```python
myFile = open("Demo.txt")

for x in myFile:
  print(x)
```

#Closing Files
Always close a file when you are done with it as good practice

```python
myFile = open("Demo.txt")
for x in myFile:
  print(x)
myFile.close()
```

#Writing to an Existing File
In order to write to an existing file use 'open()', 'a' to append a message to the end, and 'w' to write

```python
myFile = open("Demo.txt", "a")
myFile.write("Did you know Salamence learns fly upon evolution?")
myFile.close()

#Open and the read the file after the new content has been appended

myFile = open("Demo.txt", 'r')

print(myFile.read())
```

```python
#Using the write call overwrites all of the content
myFile = open("Demo.txt", "w")
myFile.write("This is my domain now punk!")

myFile = open("Demo.txt", "r")
print(myFile.read())
```

Append adds to the file while Write replaces the file with the new content you provide.

#Creating a New File
You can create a new file in Python using 'open()' and 'x' to create a new file. You can even use 'a' or 'w' to create a new one as well by appending or writing.

```python
myFile = open("DemoTwo.txt", "x")
```

```python
#Write into the new file
myFile = open("DemoTwo.txt", "w")

myFile.write("This is the first file that we have created from scratch!")

myFile = open("DemoTwo.txt", "r")

print(myFile.read())
```

#Deleting Files
In order to delete a file you need to import the OS module and run 'os.remove()'

```python
import os
os.remove("DemoTwo.txt")
```

Checking if a file exists is good practice to avoid getting an error

```python
import os
if os.path.exists("DemoTwo.txt"):
  os.remove("DemoTwo.txt")
else:
  print("This file does not exist")
```
