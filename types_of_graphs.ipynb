In this notebook we are going to go over the different types of graphs like subplots and pie charts. This notebook will conclude the basics of Math Plotting in Python

We are going to create a subplot with the subplot() function

```python
import matplotlib.pyplot as plt
import numpy as np

#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)
plt.plot(x,y)

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)
plt.plot(x,y)

plt.show()
```

The subplot has three arguments: rows, columns, and index of current plot
1 row 2 columns and 1st plot.

You can add a title to each plot with title()

```python
import matplotlib.pyplot as plt
import numpy as np

#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 2, 1)
plt.plot(x,y)
plt.title("CARS SOLD")

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 2, 2)
plt.plot(x,y)
plt.title("MONEY EARNED")
plt.show()
```

You can add a super title with suptitle()

```python
import matplotlib.pyplot as plt
import numpy as np

#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("SALES")

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)
plt.plot(x,y)
plt.title("INCOME")

plt.suptitle("MY SHOP")
plt.show()
```

To create a scatter plot you can use scatter() to draw a scatter plot. It plots one dot for each observation. It needs two arrays of the same length with one being the values for the x axis and the other being the values of y axis

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])

plt.scatter(x,y)
plt.show()
```

To compare plots draw two plots on the same figure

```python
import matplotlib.pyplot as plt
import numpy as np

#First plot:
x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
plt.scatter(x,y)

#Second Plot
x = np.array([2,2,8,1,15,8,12,9,7,3,11,4,7,14,12])
y = np.array([100,105,84,105,90,99,90,95,94,100,79,112,91,80,85])
plt.scatter(x,y)

plt.show()
```

To set the colors use the color or c argument

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x, y, color = 'hotpink')

x = np.array([2,2,8,1,15,8,12,9,7,3,11,4,7,14,12])
y = np.array([100,105,84,105,90,99,90,95,94,100,79,112,91,80,85])
plt.scatter(x, y, color = 'yellow')

plt.show()

```

You can even color each dot with c

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
colors = np.array(['red','blue','green','yellow','orange','purple','lime','pink','cyan','black','red','blue','green'])

plt.scatter(x,y, c = colors)

plt.show()
```

You can create a color map to show the range of colors like blue is cold and red is hot. You can specify the colormap with the keyword 'cmap' and then input the value of the colormap with a pre built one. You also have to create an array from 0 to 100.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
colors = np.array([0, 10, 20, 30, 40, 45, 50, 55, 60, 70, 80, 90, 100])

plt.scatter(x, y, c = colors, cmap = 'viridis')

plt.show()
```

You can include a colormap with plt.colorbar()

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
colors = np.array([0, 10, 20, 30, 40, 45, 50, 55, 60, 70, 80, 90, 100])

plt.scatter(x, y, c = colors, cmap = 'viridis')

plt.colorbar()

plt.show()
```

You can change the size of the dots with the s argument.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([5, 56, 12, 89, 45, 12, 34, 56, 12, 14, 16, 20, 19])

plt.scatter(x, y, s=sizes)

plt.show()
```

You can adjust the transparency of the dots with the alpha argument.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([5, 56, 12, 89, 45, 12, 34, 56, 12, 14, 16, 20, 19])

plt.scatter(x, y, s=sizes, alpha = 0.5)

plt.show()
```

You can create random arrays and combine color size with alpha. Try it now

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(100, size = (100))
y = np.random.randint(100, size = (100))
colors = np.random.randint(100, size = (100))
sizes = 10 * np.random.randint(100, size = (100))

plt.scatter(x,y, c = colors, s = sizes, alpha = 0.5, cmap = 'rainbow')
plt.colorbar()
plt.show()
```
