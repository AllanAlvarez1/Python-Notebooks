We are going to learn about bar graphs. In order to draw a bar graph, use the bar() function

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['A','B','C','D'])
y = np.array([1,3,5,7])

plt.bar(x,y)
plt.show()
```

bar() takes in the categories as the first parameter and the value as the second parameter in the form of arrays

```python
x = ['Apples', 'Bananas']
y = [30, 20]
plt.bar(x,y)
```

for horizontal bars, use barh()

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['Metal Gear', 'Pokemon', 'Sonic the Hedgehog', 'Persona'])
y = np.array([20, 80, 10, 30])

plt.barh(x, y)
plt.show()
```

to change the bar color use the argument 'color'

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['Red', 'Blue', 'Green', 'Purple'])
y =  np.array([30, 40, 60, 70])
colors = (['Red', 'Blue', 'Green', 'Purple'])

plt.bar(x, y, color = colors)
plt.show()
```

The argument 'width' sets the width of the bars

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['A','B','C','D'])
y = np.array([1, 5, 2, 8])

plt.bar(x,y, width = 0.1)
plt.show()
```

You can adjust the height of horizontal bars too the same way

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['A','B','C','D'])
y = np.array([3, 5, 2, 9])

plt.barh(x, y, height = 0.1)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['Aaron', 'Kameron', 'Allan', 'Elias'])
y = np.array([40, 30, 50, 40])
colors = (['Red', 'Purple', 'Green', 'Blue'])

plt.bar(x, y, color = colors, width = 0.5 )
plt.show()
```

Making a histogram is easy. Just use hist() and use an array of numbers to create the histogram. The parameters are (array of n length values, deviation, and concentrated value amount.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

Making a pie chart is easy too. Just use the pie() function and give an array for it to use

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([20, 10, 50, 39])

plt.pie(y)
plt.show()
```

It starts on the x-axis and moves counterclockwise.

# Labels
You can add labels to a pie chart with the 'labels' paramerer. It has to be an array with one label for each wedge.

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([50, 20, 20, 30])
myLabels = ["Pokemon", "Megaman", "Dark Souls", "Pac-Man"]

plt.pie(y, labels = myLabels)
plt.show()
```

#Starting Angle

You can change the starting angle by using the 'startangle' parameter. Its default is 0 but you can choose between 0 and 270 degrees

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([25, 25, 50, 25])
myLabels = ["MegaMan", "Smash Bros", "Fire Emblem", "Pikmin"]
myStartingAngle = 180

plt.pie(y, labels = myLabels, startangle = myStartingAngle)
plt.show()
```

#Explode!
If you want one of the wedges to stand out, use the 'explode' parameter. Each value represents how far from the center each wedge is displayed.

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([30, 40, 50, 90])
myLabels = ["Jojo", "One Piece", "Naruto", "Dragon Ball"]
explosion = [0.2, 0.1, 0.5, 0.3]

plt.pie(y, labels = myLabels, explode = explosion)
plt.show()
```

#Shadows
You can add shadows to a pie chart by setting the 'shadow' parameter to True

```python
import matplotlib.pyplot as plt
import numpy as np

y = ([20, 40, 60, 80])
myLabels = ["Dark Souls", "Pokemon", "Super Mario", "Sonic the Hedgehog"]

plt.pie(y, labels = myLabels, shadow = True)
plt.show()
```

#Colors
You can set the colors of each by using the 'colors' parameter

```python
import matplotlib.pyplot as plt
import numpy as np

y = ([5, 20, 30, 10])
myLabels = ["New York", "Albany", "Florida", "Miami"]
myColors = ["cyan", "pink", 'yellow', 'purple']

plt.pie(y, labels = myLabels, colors = myColors, shadow = True)
plt.show()
```

#Legends
You can add a list to explain each wedge by using the 'legend()' function.
You can even use give it a header using the 'title' paramater

```python
import matplotlib.pyplot as plt
import numpy as np

y = ([30, 40, 70, 10])
myLabels = ["Cobra Kai", "Sly Cooper", "Megaman", "Jenny"]
myColors = ["b", "c", "m", "g"]

plt.pie(y, labels = myLabels, colors = myColors, shadow = True)
plt.legend(title = "Four Legends")
plt.show()
```
